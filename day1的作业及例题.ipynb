```python
#例题
# 定义一家企业的基本信息
company_name = "华能制造有限公司"      # 字符串
monthly_electricity = 50000           # 整数：月用电量（度）
grid_factor = 0.5703                  # 浮点数：电网排放因子

# 输出信息
print("企业名称：", company_name)
print("月用电量：", monthly_electricity, "度")
print("电网排放因子：", grid_factor, "吨CO2/万度")

# 计算碳排放量（公式：用电量 × 排放因子 / 10000）
co2 = monthly_electricity * grid_factor / 10000
print("该企业本月碳排放：", co2, "吨CO2")
```

```python
# 作业
# 碳中和场景：存储企业的能耗数据和排放因子
# 定义一家企业的天然气用量（2000立方米）、天然气排放因子（2.162吨CO2/千立方米），计算并输出天然气产生的碳排放。#
company_name = "我承认这是幻觉"
gas = 2000
gas_factor = 2.162

# 输出信息
print("企业名字：",company_name)
print("天然气用量：",gas,"立方米")
print("天然气排放因子：",gas_factor,"吨CO2/千立方米")

# 计算碳排放量 （公式：天然气用量 * 排放因子/1000）
co2 = gas * gas_factor/1000
print("天然气产生的碳排放：",co2,"吨")
```
