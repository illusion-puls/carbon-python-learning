```python
#对第一次小游戏的改进#
#1、如果玩家猜错了需要给点提示#
#2、需要给玩家多刺试验机会#
#3、使用随机数替换固定答案#
import random
answer = random.randint(1,10)

counts = 3

while counts > 0:
    temp = input("不妨猜猜小甲鱼现在心中想的是哪一个数字？你只有三次机会，要好好把握哟！")
    guess = int(temp)
    if guess == answer:
        print("你是小甲鱼肚子里的蛔虫吗？")
        print("哼，猜中了也没有奖励！")
        break
    else:
#这里出现过一个bug，guess没有和answer比较，而是和上次没有改过来的8比较，#
#于是出现的一个滑稽的现象，answer生成8，但是输入8却不算正确答案，#
#使得玩家认为answer为一个存在于8和9之间的整数#
        if guess > answer:
            print("大啦~")
        else:
            print("小啦~")
        counts = counts - 1
        
print("游戏结束，不玩了：）")
```
